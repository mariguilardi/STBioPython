# **Pandas Dataframe**

É a principal estrutura semelhante à planilhas de Excel/ Google Sheet.
- É uma estrutura de dados **tabular bidimensional** mutável em tamanho
- Pode armazenar diferentes tipos de dados (int, float, string, date e etc.),
- Possui eixos rotulados (linhas e colunas)

<img style="float: center;" src="https://external-content.duckduckgo.com/iu/?u=https%3A%2F%2Frwegener2.github.io%2Fsarp_lessons%2F_images%2Fpandas_dataframe.png&f=1&nofb=1&ipt=0c349e8c8c8b9d9b997930dbcccece110f7ef88e5bc885e7b3842d0a4bd19e08&ipo=images" width="500"/>

## **Series**

*Series* é uma matriz rotulada unidimensional capaz de conter qualquer tipo de dados (inteiros, strings, decimais, etc.)

Cada coluna no *dataframe* é uma series, é igual uma coluna no Excel/ Google Sheet.

O índice começa no zero (como em strings e listas).

<img style="float: center;" src="https://external-content.duckduckgo.com/iu/?u=https%3A%2F%2Fdatascientyst.com%2Fcontent%2Fimages%2F2021%2F12%2Fpandas-series-index-values.png&f=1&nofb=1&ipt=a45ececfad8d981be231a04f055a1998c6b5c9915e03b77739f53f52cf01db50&ipo=images" width="500"/>


````python
# transformando nossa lista anterior em uma series
nomes = pd.Series(lista)
print(nomes)
````

````python
# criando uma serie com índice texto
alunos = pd.Series(lista, index=["a", "b", "c"],  name = "nomes")
alunos
````

# **Seleção e filtragem de dados**



## **Seleção por nome de coluna**

Selecionar colunas por rótulo (nome da coluna) utilizando .loc

`<nome dataframe>.loc["nome da coluna"]`

<img style="background-color: white;" src="https://pandas.pydata.org/docs/_images/03_subset_columns.svg"/>


````python
# selecionar uma coluna por nome (ou seja, uma series) - uma dimensão
pinguim.loc[:,"Data_ovo"]
````

````python
# selecionar múltiplas colunas por nome - duas dimensões
pinguim.loc[:,["Especie", "Sexo", "Data_ovo"]]
````

````python
# selecionar um intervalo de colunas
pinguim.loc[:, "Comprimento_bico":"Comprimento_asa"]
````

## **Seleção de linhas por nome da linha**

Selecionar linhas por nome da linha utilizando .loc

`<nome dataframe>.iloc[nome da linha,]`

<img style="background-color: white;" src="https://pandas.pydata.org/docs/_images/03_subset_rows.svg"/>

````python
pinguim.head()
````

````python
# .loc - selecionar uma linha utilizando nome da linha como Series
# uma dimensão
pinguim.loc["P1", :]
````

````python
# .loc - selecionar uma linha utilizando nome da linha como Dataframe
# duas dimensões
pinguim.loc[["P1"], :]
````

````python
# .loc - selecionar múltiplas linhas com índices
pinguim.loc["P2":, :]
````

## **Selecionar linhas e colunas**

Selecionar colunas e linhas por índice e nome de colunas utilizando .loc

`<nome dataframe>.iloc[índice, "nome de coluna"]`

<img style="background-color: white;" src="https://pandas.pydata.org/docs/_images/03_subset_columns_rows.svg"/>


````python
# selecionar valor único
pinguim.loc["P1","Especie"]
````

````python
# selecionar linhas e colunas
pinguim.loc["P1":"P10","Especie":"Data_ovo"]
````

````python
# selecionar linhas e algumas colunas
pinguim.loc["P1":"P10",["Especie", "Ilha", "Sexo"]]
````

## Seleção po índice

Da mesma forma como em *strings* e lista, é possível selecionar por índice utilizando .iloc

`<dataframe>.iloc[<número índice>]`

````python
pinguim.head()
````

````python
# selecionar uma linha passando um índice (número inteiro)
pinguim.iloc[1]
````

````python
# selecionar por posição de linha e posição de coluna com .iloc
pinguim.iloc[2, 0]
````

````python
# selecionar por posição de uma coluna com .iloc - Series
# uma dimensão
pinguim.iloc[:,0]
````

````python
# selecionar por posição de múltiplas colunas com .iloc
# duas dimensões
pinguim.iloc[:,1:3]
````

# **Filtrar dados**

## **Filtro com operadores lógicos**

- Igual a: ==
- Diferente de: !=
- Maior que: >
- Menor que: <
- Maior ou igual que: =>
- Menor ou igual que: <=

````python
# usar operadores de comparações para filtrar informações específicas
pinguim["Sexo"] == "Fêmea"
````

````python
# associar filtro a variável
filtro_femea = pinguim['Sexo'] == "Fêmea"
````

````python
# selecionar os dados no dataframe usando o filtro
pinguim[filtro_femea]
````

````python
# ou declare o filtro diretamente na seleção
pinguim[pinguim['Sexo'] == "Fêmea"]
````

## **Filtro com múltiplos operadores**

- E: `&`
- Ou: `|`

````python
# criar filtros
femea = pinguim["Sexo"] == "Fêmea"
sp = pinguim["Especie"] == "Pygoscelis adeliae"
````

````python
# filtro com operadores de comparação e lógicos
pinguim[femea & sp]
````

# **Criar colunas e tabelas a partir dos próprios dados**



## **Criar colunas usando operações aritméticas**

````python
# criando coluna com a soma de duas variáveis
pinguim['Massa_corporal_corrigida'] = pinguim['Massa_corporal'] + 100
pinguim.head(2)
````

````python
# criando coluna com a multiplicação de duas variáveis
pinguim["Peso_kg"] = pinguim["Massa_corporal"] / 1000
pinguim.head(3)
````

## **Criar colunas usando operações com *strings***

````python
# separando uma coluna em várias utilizando a coluna Data_ovo
pinguim.loc[:,["Especie","Data_ovo"]]
````

````python
# separando Data_ovo em dia, mês e ano
split_data = pinguim["Data_ovo"].str.split("/", n = 2, expand = True)
split_data
````

````python
# criando nova coluna com mês
pinguim["Mês"] = split_data[1]
pinguim["Ano"] = split_data[2]
pinguim.head()
````

# **Excluindo colunas e linhas**

É possível excluir colunas e linhas usando `drop(`.

`<nome do dataframe>.drop(columns = <nome da coluna>, axis = 1)`


````python
# quantas colunas temos?
pinguim.shape
````

````python
# excluindo a coluna Data_ovo
pinguim.drop(columns = "Data_ovo").head(2)
````

````python
# excluindo colunas
pinguim.drop(columns=['Massa_corporal_corrigida', 'Peso_kg']).head(2)
````

````python
pinguim.shape
````

````python
# excluindo linhas usando índice
# excluindo a primeira linha
pinguim.drop("P1")
````

````python
# quantas colunas ficamos?
pinguim.shape
````

````python
# também é possível excluir linhas usando filtro
# quais os valores em ano?
pinguim["Ano"].unique()
````

````python
# criamos um filtro que exclui ano de 2007
filtro_ano = pinguim["Ano"] != "2007"
````

````python
# filtramos e adicionamos a novo dataframe
pinguim_ano = pinguim[pinguim["Ano"] != 2007]
pinguim_ano.shape
````

# **Checando valores ausentes**

É possível checar valores ausentes (ou *missing data*) usando a funçao is.na().

O Python representa valores ausentes como `NA`.

`<nome dataframe>.isna()`



````python
# valores ausentes no dataframe
pinguim.isna()
````

````python
# contando número de linhas com valores ausentes numa coluna
pinguim["Sexo"].isna().sum().item()
````

````python
pinguim.shape
````

````python
# retirando linha com valores ausentes
pinguim_limpo = pinguim.dropna()
````
