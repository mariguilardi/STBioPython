### **Boas Práticas de Nomes de variáveis**

- Devem começar com uma letra e podem conter letras, números, _ e .

- Sugere-se usar *snake_case*, ou seja, palavras escritas em minúsculo separadas pelo *underscore* (_). Por exemplo:

```
alunos_python <- "40"
```

O Python é *case sensitive*, isto é, faz a diferenciação entre as letras minúsculas e maiúsculas. Portanto, uma variável chamado ***a***lunos é diferente de uma outra variável chamada ***A***lunos.
