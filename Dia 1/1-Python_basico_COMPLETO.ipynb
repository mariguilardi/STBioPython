### **Boas Práticas de Nomes de variáveis**

- Devem começar com uma letra e podem conter letras, números, _ e .

- Sugere-se usar *snake_case*, ou seja, palavras escritas em minúsculo separadas pelo *underscore* (_). Por exemplo:

```
alunos_python = "40"
```

O Python é *case sensitive*, isto é, faz a diferenciação entre as letras minúsculas e maiúsculas. Portanto, uma variável chamado ***a***lunos é diferente de uma outra variável chamada ***A***lunos.

## **Lógicos**

***Boolean*** são valores de `True` (verdadeiro) ou `False` (falso)

`booleano = True`

**Operadores lógicos ou relacionais:** servem para fazer perguntas que possam ser
respondidas com True ou False (verdadeiro/falso)

- Maior que: `>`
- Menor que: `<`
- Maior ou igual a: `>=`
- Menor ou igual a: `<=`
- Idêntico: `==`
- Diferente de: `!=`
- Não: `not`
- E: `and` (todas as condições precisam ser satisfeitas)
- Ou: `or` (apenas uma das condições precisa ser satisfeita)

````python
# vamos declarar três variáveis diferentes
a = 2
b = 3
c = 2
````

````python
# igual a
a == b
````

````python
# menor ou igual a
a <= b
````

````python
# menor que e menor que
a < b and b < c
````

````python
# menor que ou menor que
a < b or b < c
````

## **Listas**

- permitem armazenar várias informações diferentes (número, *string*, lógico) em uma mesma variável.


````python
# criando uma lista com tipos diferentes
curso = ["dados", True, 2026]
print(curso)
````

````python
# criando uma lista com mesmos tipos
evento = ["dados", "flora", "fauna"]
print(evento)
````

````python
# quantos itens tem na lista evento?
len(evento)
````

````python
## fatiando listas
curso[0]
````

````python
# e listas podem estar dentro de listas
semana = ["Python", ["morfologia", "geométrica"], "lianas"]
print(semana)
````

````python
# como fica quando acessamos a lista?
semana[1]
````

````python
# e como acessamos dentro dessa lista na lista?
semana[1][0]
````

````python
# adicionando elementos na lista utilizando .append() sem criar nova variável
semana.append("arqueobotânica")
````

````python
print(semana)
````

````python
# separar string em elementos de uma lista usando .split()
anos = "2024,2025,2026"
print(anos)
````

````python
anos.split(",")
````

````python
lista_anos = anos.split(",")
print(lista_anos)
````

## **Tupla**

São similares às listas, mas **imutáveis**.

Não podemos adicionar ou modificar nenhum de seus elementos.

````python
# criando uma tupla
codons = ("AUG", "AGG", "UAA")
print(codons)
````

````python
# acessando elementos na tupla
codons[1]
````

````python
# e se adicionarmos um novo elemento?
codons.append("AUG")
````
